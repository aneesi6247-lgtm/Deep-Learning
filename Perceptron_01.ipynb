```python
import torch
import torch.nn as nn
```

```python
# input
inputs = torch.tensor([1.0,2.0,3.0])
# tensor are basically multidimmentional arrays that we can create in pytorch
torch.manual_seed(42) # to get the same results each time
neuron = nn.Linear(in_features=3, out_features=1)
output = neuron(inputs)
output
```

```python
# to verify the output
neuron.weight @ inputs + neuron.bias
```
